```python
# Setup Repository and Environment

import os
import shutil
import subprocess
from pathlib import Path

repo_path = Path("/kaggle/temp/mnist_active_screening")
output_path = Path("/kaggle/working/runs")

project_files = list(Path("/kaggle/input").rglob("pyproject.toml"))

if len(project_files) != 1:
    raise RuntimeError(
        f"Expected exactly one project source, found: {project_files}"
    )

source_path = project_files[0].parent

shutil.copytree(
    source_path,
    repo_path,
    dirs_exist_ok=True
)

output_path.mkdir(parents=True, exist_ok=True)

runs_path = repo_path / "runs"

if runs_path.is_symlink() or runs_path.is_file():
    runs_path.unlink()
elif runs_path.is_dir():
    shutil.rmtree(runs_path)

os.symlink(output_path, runs_path)

subprocess.run(
    ["python", "-m", "pip", "install", "-e", f"{repo_path}[rl]"],
    check=True
)

print(f"Source loaded from: {source_path}")
print("Environment ready.")
print(f"Results will be saved to: {output_path}")
```
